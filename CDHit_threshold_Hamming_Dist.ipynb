#Notes
NNK = 4x4x2 = 32 possibilities
library size = 1.5 E8 
theoretical library size = 32^9
Number of neighbors 2bp away from 1 sequence: 1899 = sum for i=[0,8]: 3(7i+4)+3(7i+1)+7i
Number of neighbors 1bp away from 1 sequence: 62 = 9(3+3+1) = 7x9

# 2bp away only

```julia
p1 = (7*9*(1/7*62+6/7*60)/2)/(32^9) #(number of possible 2bp away neighbors from a peptide)/(number of possible peptides)
 #probability of 2bp neighbors for a given sequence in all possible peptides
```

```julia
n = 1.5*1E8 #library size
E = n*(1-(1-p1).^(n-1)) #expected number of 2bp-away-neighbors in library
```

# 1bp away only

```julia
p2 = (7*9)/(32^9) #(number of possible 1bp away neighbors from a given peptide)/(number of possible peptides)
```

# 1bp and 2bp away 
## multiple ways of calculating give similar/same values

```julia
p = (7*9*(1/7*62+6/7*60)/2 + 7*9)/(32^9) #(number of possible 1bp and 2bp away neighbors from a given peptide)/(number of possible peptides)
```

```julia
Et = n*(1-(1-p).^(n-1)) # Expected total number of sequences 1 or 2bp away from another sequence
```

```julia
P12 = p1+p2 # probability of a peptide being 1 or 2 bp apart from a given peptide
```

```julia
E12 = n*(1-(1-P12).^(n-1)) #same as Et (calculated slightly differently)
```

```julia
E1 = n*(1-(1-p1).^(n-1)) # Total num of sequences 2bp away
E2 = n*(1-(1-p2).^(n-1)) # Total num of sequences 1bp away
```

```julia
Ee = E1 + E2 - E1 * E2/n # Expected num of sequences 1 or 2bp away (accounting for overlap)
```

```julia
PA = 1-(1-p2)^(n-1) # Prob of a sequence being 1 bp away from any sequence in the lib
```

```julia
PB = 1-(1-p1)^(n-1) # Prob of a sequence being 2 bp away from any sequence in the lib
```

```julia
PAB = PA+PB-PA*PB # Prob of a sequence 1 or 2 bp away from any other sequence
```

```julia
EAB = n*PAB # Same as Ee
```
